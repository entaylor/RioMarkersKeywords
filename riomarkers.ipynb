```python
# specify the necessary inputs:
# the oecd data file that we will search through
oecd_data_filename = 'CRS.parquet'
# the list of keywords to search for
keyword_data_filename = 'keywords_list.txt'

# at this stage we only want to initialise;
# we will step through the process in the cells below.
init_only = True

debug = 1 # use debug = 9 for maximum verbosity

# initialise, load, and setup for the process
self = data = riomarkers.OECDDataset(
    oecd_data_filename=oecd_data_filename,
    oecd_data_filename_list=None,
    oecd_data_separator=None,
    oecd_data_dtypes=None,
    keyword_data_filename=keyword_data_filename,
    debug=debug, init_only=init_only)
```

```python
# next we prepare the data:
#    - deal with missing data
#    - convert all text to lowercase
#    - replace special characters to plain ascii
#    - join title and description to search both

oecd_title_colname='project_title'
oecd_desc_colname='long_description'
self.prepare_oecd_data(
    oecd_title_colname=oecd_title_colname,
    oecd_desc_colname=oecd_desc_colname,
)
```
